```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if "outcome" in num_cols:
    num_cols = [c for c in num_cols if c != "outcome"]
corr = df[num_cols].corr()
plt.figure(figsize=(12, 10))
sns.heatmap(corr, cmap="coolwarm", center=0, annot=False, square=True)
plt.title("Correlation matrix (numeric predictors)")
plt.tight_layout()
plt.show()
```

**Multicollinearity & feature pruning:** The heatmap above shows strong correlations among some predictors (e.g. x, y, z; price and carat). Low-signal features (e.g. a5–a9, b5–b9) were tested for removal in **03_tree_models.ipynb**; dropping a subset of them improved CV R² (~0.477 → ~0.480). Use that experiment and this heatmap in the report.
